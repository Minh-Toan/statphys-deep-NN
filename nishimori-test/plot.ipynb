```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['text.usetex'] = True
```

```python
width, height = 16, 4
size = int(width*1.5)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(width, height), tight_layout=True)

ax1.tick_params(axis='both', labelsize=size)
ax1.tick_params(axis='both', labelsize=size)
ax1.set_xlabel(r'HMC step', fontsize=size )
ax1.grid()
# ax1.set_xlim(0, 510)
# ax1.set_ylim(0, 0.22)

ax2.tick_params(axis='both', labelsize=size)
ax2.tick_params(axis='both', labelsize=size)
ax2.set_xlabel(r'HMC step', fontsize=int(1.2*size))
ax2.grid()
# ax2.set_ylim(0, 0.031)
# ax2.set_xlim(0, 5000)


dev_unfo = np.loadtxt('dev_unfo.csv', delimiter=',')
ax1.plot(dev_unfo, label='uninformative HMC')
ax1.legend(fontsize=int(size))


dev_info = np.loadtxt('dev_info.csv', delimiter=',')
idx = np.linspace(0, len(dev_info) - 1, 500, dtype=int)
ax2.plot(idx, dev_info[idx], color='red', label='informative HMC')
ax2.legend(fontsize=int(size))

plt.savefig('metastable_confirmation.pdf')
```
