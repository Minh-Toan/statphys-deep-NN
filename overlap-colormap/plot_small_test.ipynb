```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
alpha = 2.75
nstep = 250000
qw1s = np.loadtxt(f'small_test_data/qw1_alpha_{alpha}_step_{nstep}.csv', delimiter=',').flatten()
qw2s = np.loadtxt(f'small_test_data/qw2_alpha_{alpha}_step_{nstep}.csv', delimiter=',').flatten()

```

```python
plt.plot(qw1s, label='W_1 overlap')
plt.plot(qw2s, label='W_2 overlap')
plt.legend()
plt.grid()
plt.show()
```
