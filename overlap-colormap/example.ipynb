```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
```

```python
vmin, vmax = -2, 2
m = 100

v1s = np.linspace(vmin, vmax, m+1)
v2s = np.linspace(vmin, vmax, m+1)
V1, V2 = np.meshgrid(v1s, v2s)

A = np.random.randn(m,m)
B = np.random.randn(m)
C = np.random.randn(m)

B = B[:, np.newaxis]
C = C[np.newaxis, :]
```

```python
cmap = 'viridis'
fig = plt.figure(figsize=(5.5, 5))
gs = GridSpec(nrows=2, ncols=3, figure=fig, width_ratios=[0.15, 1, 0.1], height_ratios=[1, 0.15], hspace=0.1, wspace=0.1)

ax_main = fig.add_subplot(gs[0, 1])
im_main = ax_main.pcolormesh(V2, V1, A, cmap=cmap)
ax_main.set_xlim(vmin, vmax)
ax_main.set_ylim(vmin, vmax)
ax_main.set_xticklabels([])
ax_main.set_yticklabels([])

ax_left = fig.add_subplot(gs[0, 0])
im_left = ax_left.pcolormesh([0, 1], v1s, B, cmap=cmap)
ax_left.set_ylabel(r'$v$')
ax_left.set_title(r'$Q_{2;1}(v)$', fontsize=10, pad=5, loc='left')
ax_left.set_xticks([])
ax_left.set_ylim(vmin, vmax)

ax_bottom = fig.add_subplot(gs[1, 1])
im_bottom = ax_bottom.pcolormesh(v2s, [0, 1], C, cmap=cmap)
ax_bottom.set_xlabel(r'$v^{(2)}$')
ax_bottom.set_yticks([])
ax_bottom.set_xlim(vmin, vmax)
ax_bottom.set_ylabel(r'$Q_1(v^{(2)})$', fontsize=10)

ax_cbar = fig.add_subplot(gs[0:2, 2])
fig.colorbar(im_main, cax=ax_cbar)

plt.show()
```
