```python
L = 2
alpha = 3.5
Delta = 0.1

p = lambda q: potential(alpha, Delta, q)

# x = np.concatenate((np.linspace(0, 0.95, 21), np.linspace(0.95, 0.999, 21)[1:]) )
# y = np.concatenate((np.linspace(0, 0.95, 21), np.linspace(0.95, 0.999, 21)[1:]) )

x = np.linspace(0.99, 0.999, 21)
y = np.linspace(0.95, 0.999, 21)

X, Y = np.meshgrid(x, y)

l = len(x)
Z  = np.zeros((l, l))
for i in range(l):
    for j in range(l):
        Z[i,j] = p(np.array([X[i,j], Y[i,j]]))
        

plt.pcolormesh(X, Y, Z, cmap='coolwarm')
plt.colorbar()
# plt.axis('equal')
plt.show()
```
