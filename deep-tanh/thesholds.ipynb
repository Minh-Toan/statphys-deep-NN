```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import AAA
from deep_tanh import *
```

```python
def bisect(f,a,b,eps=1e-6): # find root of f, f monotone on (a, b)
    while b-a>eps:
        m=(a+b)/2
        a,b=(a,m) if f(a)*f(m)<=0 else (m,b)
    return (a+b)/2

def bisect_2(f,a,b,eps=1e-6): # find smallest x such that f becomes nonzero
    while b-a>eps:
        m=(a+b)/2
        if f(m)>0.001:
            a, b = a, m
        else:
            a, b = m, b
    return (a+b)/2
    
```

```python
'''
threshold for informative overlap > 0
'''

Delta = 0.1
Ls = [1, 2, 3, 4, 5]
bounds = np.array([ [1,2], [3,4], [5, 6], [6,7], [8,9] ])
alphas_c = []

for L, bound in zip(Ls, bounds):
    def _mmse(alpha):
        return np.nan_to_num(1-quantities(np.ones(L), alpha, Delta)[0], nan=0.0)
    alpha_c = bisect_2(_mmse, bound[0], bound[1])
    alphas_c.append(alpha_c)
alphas_c
```

```python
'''
threshold for equilibrium overlap > 0
'''

bounds = np.array([[1.5, 3], [4,6], [8, 10], [14, 16], [20, 21]])

Ls = [1, 2, 3, 4, 5]
Delta = 0.1
alphas_e = []
mmses_e = [] # generalisation error at theses threshold
for L, bound in zip(Ls, bounds):
    def pot_diff(alpha): # comparison with potential at 0
        q_0 = np.zeros(L)
        q_info = quantities(np.ones(L), alpha, Delta)[1:]
        return potential(alpha, Delta, q_info) - potential(alpha, Delta, q_0)
    alpha_e = bisect(pot_diff, bound[0], bound[1])
    mmse_e = quantities(np.ones(L), alpha_e, Delta)[0]
    alphas_e.append(alpha_e)
    mmses_e.append(mmse_e)
```
