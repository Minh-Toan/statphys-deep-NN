```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import AAA
```

```python
# AAA approximation of g
qs = np.linspace(0, 1, 21)
gs = np.loadtxt('g_21.csv', delimiter=',')
g = AAA(qs, gs)
```

```python
qs = np.linspace(0, 1, 201)
gs = np.loadtxt('g_201.csv', delimiter=',')
plt.plot(qs, np.abs(gs-g(qs)))
plt.yscale('log')
plt.show()
```
