```python
import numpy as np
import func
```

## $\tanh(2x)$ normalized

```python
sig = lambda x: np.tanh(2*x)
nu = func.Ez( lambda x: sig(x)**2 )
np.sqrt(nu)
```

## ReLU with $\mu_1$ removed

```python
sig = lambda x: np.maximum(x, 0)
mu0 = func.Ez( lambda x: sig(x) )
mu1 = func.Ez( lambda x: x*sig(x) )
mu0, mu1
```

## $\tanh(2x)$ with $\mu_1$ removed and normalized

```python
sig = lambda x: np.tanh(2*x)

mu1 = func.Ez( lambda x: x*sig(x) )
nu = func.Ez( lambda x: sig(x)**2 )
sig_n = lambda x: (sig(x)-mu1*x)/np.sqrt(nu - mu1**2) # normalized sig
```

```python
mu1, np.sqrt(nu - mu1**2)
```

```python
m = 201
qs = np.linspace(0, 1, m)
gs = np.array([func.h(q, sig_n) for q in qs])
np.savetxt(f'g_{m}.csv', gs, delimiter=',')
```
