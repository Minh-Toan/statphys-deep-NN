```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
import matplotlib
matplotlib.rcParams['text.usetex'] = True

import func
```

```python
def cov(X):
    X = X.reshape(len(X), -1)
    _, C = func.normalize(X)
    return C
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train_small = func.downsize_mnist(x_train)
x_test_small = func.downsize_mnist(x_test)
eig_train_small = np.linalg.eigvalsh(cov(x_train_small))
eig_test_small = np.linalg.eigvalsh(cov(x_test_small))
```

```python
width, height = 4, 5
size = int(width*2.4)
fig, ax = plt.subplots(figsize=(width, height), tight_layout=True)
ax.tick_params(axis='both', labelsize=size)
ax.set_xticks([0, 5, 10, 15, 20])
ax.set_yticks([0, 0.4, 0.8, 1.2, 1.6])
ax.grid()

# ax.hist(eig_train_small, bins=40, density=True, facecolor='black', edgecolor='white', label='train', alpha=0.7)
# ax.hist(eig_test_small, bins=40, density=True, facecolor='red', edgecolor='white', label='test', alpha=0.7)

bins = np.linspace(0, 20, 41)

counts_train, _ = np.histogram(eig_train_small, bins=bins, density=True)
ax.hist(eig_train_small, bins=bins, density=True, facecolor='black', edgecolor='white', label='train', alpha=0.7)

bin_centers = 0.5 * (bins[:-1] + bins[1:])
counts_test, _ = np.histogram(eig_test_small, bins=bins, density=True)
for x, y in zip(bin_centers, counts_test):
    ax.vlines(x, 0, y, colors='red', lw=2, label='test' if x == bin_centers[0] else "")



ax.legend(fontsize=size)
ax.set_xlim(0, 20)
# ax.set_ylim(0, 1.6)

plt.savefig('spectral.pdf')
```
